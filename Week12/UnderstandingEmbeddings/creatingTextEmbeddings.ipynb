```python
# import numpy as np

import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Get embeddings for two different texts
text1 =  "Pennsylvania is home to a thriving population of American black bears (Ursus americanus), with current estimates suggesting around 18,000 bears inhabit the state's diverse forests and mountains. These adaptable omnivores, which can weigh between 200-600 pounds when fully grown, are primarily found in the state's northern and central regions, particularly in the Pocono Mountains and across the Allegheny National Forest. Pennsylvania black bears typically maintain a diet consisting of berries, nuts, insects, and occasionally small mammals, with their feeding habits intensifying during fall as they prepare for winter dormancy. The Pennsylvania Game Commission carefully manages the bear population through regulated hunting seasons, which help control numbers while generating valuable conservation funding. Despite their intimidating size, black bears are generally shy and avoid human contact, though increasing development has led to more human-bear encounters in suburban areas, prompting wildlife officials to emphasize proper food storage and trash management to reduce potential conflicts. The black bear remains an iconic symbol of Pennsylvania's wilderness, representing both the state's rich biodiversity and its commitment to wildlife conservation."
text2 =  "Python stands as one of the most versatile and accessible programming languages in the modern software development landscape, combining remarkable simplicity with powerful capabilities that have fueled its meteoric rise in popularity. Created by Guido van Rossum and first released in 1991, Python's philosophy emphasizes code readability with its clean syntax and significant whitespace, allowing developers to express concepts in fewer lines of code than languages like C++ or Java. The language's extensive standard library, often described as providing batteries included functionality, offers ready-made modules for everything from web development and data analysis to artificial intelligence and scientific computing. Python's interpreter-based nature enables rapid prototyping and development across multiple platforms, while its dynamic typing and automatic memory management reduce common programming errors. The vibrant Python ecosystem includes frameworks like Django and Flask for web development, NumPy and Pandas for data science, and TensorFlow and PyTorch for machine learning, making it the language of choice for diverse applications ranging from simple scripts to complex enterprise systems. This combination of accessibility for beginners and depth for experts has established Python as a cornerstone technology in education, research, and industry, with a thriving community that continues to expand its capabilities through an ever-growing collection of third-party packages and libraries." 

response1 = client.embeddings.create(
    input=text1,
    model="text-embedding-ada-002"
)
embedding1 = response1.data[0].embedding

response2 = client.embeddings.create(
    input=text2,
    model="text-embedding-ada-002"
)
embedding2 = response2.data[0].embedding

# Convert to numpy arrays
embedding1_np = np.array(embedding1).reshape(1, -1)
embedding2_np = np.array(embedding2).reshape(1, -1)

# Calculate cosine similarity
similarity = cosine_similarity(embedding1_np, embedding2_np)[0][0]
print(f"Similarity between the two texts: {similarity:.4f}")
```
