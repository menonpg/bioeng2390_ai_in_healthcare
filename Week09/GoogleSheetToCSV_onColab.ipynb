```python
# prompt: load data from google sheet into a pandas dataframe

from google.colab import auth
auth.authenticate_user()

import gspread
from google.auth import default
creds, _ = default()

gc = gspread.authorize(creds)

# Open the Google Sheet by its title
spreadsheet_title = 'Purple-WhatIsPurple'  # Replace with your sheet's title
worksheet_title = 'Sheet1'  # Replace with your worksheet's title
sheet = gc.open(spreadsheet_title).worksheet(worksheet_title)

# Get all values from the sheet
data = sheet.get_all_values()

# Convert the data to a Pandas DataFrame
import pandas as pd
df = pd.DataFrame(data[1:], columns=data[0])  # Assuming the first row is a header

df =  df.dropna()

print(df.shape)
df.head()
```

```python
# write to csv file
df.to_csv('data.csv', index=False)
```
