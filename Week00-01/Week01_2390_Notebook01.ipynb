# Understanding outliers

```python
# detect the outliers in a numerical column
import numpy as np

def detect_outliers(data, margin):
    outliers_inds = []
    q1 = np.quantile(data, 0.25)
    q3 = np.quantile(data, 0.75)
    iqr = q3 - q1  # Interquartile range
    fence_low = q1 - margin * iqr
    fence_high = q3 + margin * iqr
    for ind, value in enumerate(data):
        if value > fence_high or value < fence_low:
            outliers_inds.append(ind)
    return outliers_inds


# Z score normalize
def normalize(data):
    mean = np.average(data, axis=0)
    std = np.std(data, axis=0)
    data = (data - mean) / std
    return data


# max-min normalize
def normalize2(data):
    min = np.min(data)
    max = np.max(data)
    data = (data - min) / (max - min)
    return data

outlierList = detect_outliers(df['value'].values, margin = 2)
dfClean = df.drop(outlierList)

dfClean.boxplot(column=['value'], by = "GT")
```

```python
normalizedValue_fromPython01 = normalize(df["value"].values)
df['normalizedValue_fromPython01'] = normalizedValue_fromPython01

normalizedValue_fromPython02 = normalize2(df["value"].values)
df['normalizedValue_fromPython02'] = normalizedValue_fromPython02
df.head()
```

```python
df_Normal = df[df['GT']=='Normal'].reset_index(drop=True)
df_Normal = df_Normal.drop(detect_outliers(df_Normal['value'].values, margin = 2))
df_Normal.head()
```

```python
df_Seizure = df[df['GT']=='Seizure'].reset_index(drop=True)
df_Seizure = df_Seizure.drop(detect_outliers(df_Seizure['value'].values, margin = 2))
df_Seizure.head()
```

```python
import pandas as pd
dfAll = pd.concat([df_Normal, df_Seizure])
dfAll.head()
```
